# Task 1.3: What the Paper Claims to Improve

* **Main Baseline:** SVMperf and Pegasos.

* **Limitation:** SVMperf's cutting-plane method can be slower for high-precision solutions. Pegasos (SGD) requires many iterations for small duality gaps.

* **Proposed Improvement:** Dual Coordinate Descent method that updates one variable at a time, achieving $O(l)$ scalability and faster high-precision convergence.

* **Non-Outperformance Scenario:** Low-precision regime on extremely massive datasets where one pass of SGD is enough for 'good' accuracy.
