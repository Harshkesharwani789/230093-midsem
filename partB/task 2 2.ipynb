# Task 2.2: Reproduction of DCD Algorithm

Implementing Algorithm 1 from Fan et al. (2008).

```
import numpy as np

def dcd_solver(X, y, C, eps=0.01, max_iter=100):
    l, n = X.shape
    alpha = np.zeros(l)
    w = np.zeros(n)
    Q_diag = np.sum(X**2, axis=1)
    for _ in range(max_iter):
        indices = np.arange(l)
        np.random.shuffle(indices)
        max_pg = 0
        for i in indices:
            G = y[i] * np.dot(w, X[i]) - 1
            pg = G
            if alpha[i] == 0: pg = min(0, G)
            elif alpha[i] == C: pg = max(0, G)
            max_pg = max(max_pg, abs(pg))
            if abs(pg) > 1e-12:
                oa = alpha[i]
                alpha[i] = min(max(alpha[i] - G / Q_diag[i], 0), C)
                w += (alpha[i] - oa) * y[i] * X[i]
        if max_pg < eps: break
    return w

X_train = np.load('data/X_train.npy')
y_train = np.load('data/y_train.npy')
w = dcd_solver(X_train, y_train, C=1.0)
print('Training complete.')
```
