# Task 1.2: Key Assumptions

### Assumption 1
**Assumption:** The data is high-dimensional and sparse enough that a linear boundary is sufficient.
**Why the method needs it:** LIBLINEAR relies on the efficiency of linear kernels. Coordinate descent in the dual becomes computationally expensive if the Hessian matrix cannot be calculated efficiently.
**Violation scenario:** Low-dimensional datasets with complex, non-linear boundaries like XOR.
**Paper Reference:** Section 1 (Introduction).

### Assumption 2
**Assumption:** The bias term can be represented by augmenting instances with a constant feature.
**Why the method needs it:** Algorithm 1 assumes no equality constraint $\sum y_i \alpha_i = 0$. Removing this is done by augmentation.
**Violation scenario:** Scenarios where bias regularization must differ from feature regularization.
**Paper Reference:** Appendix A.

### Assumption 3
**Assumption:** Random permutation of coordinates ensures convergence speed.
**Why the method needs it:** Deterministic order can lead to slow convergence in quadratic programming.
**Violation scenario:** Highly structured datasets where sequential updates lead to cycles.
**Paper Reference:** Algorithm 1.
